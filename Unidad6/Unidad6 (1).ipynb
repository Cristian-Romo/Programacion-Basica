```python
''' Realiza los siguientes programas y súbelos a github.
1. al usuario una temperatura en Celsius (como string) y conviértela a float. Luego calcula y muestra la temperatura en Fahrenheit con esta fórmula:
Fahrenheit = (Celsius × 9/5) + 32

2. Pide un número al usuario (puede ser entero o decimal), conviértelo a string y muestra cuántos caracteres.
''' 
```

```python
print("Bienvenido vamos a calcular la temperatura de Celsius a Fahrenheit")
# Se pide la temperatura en Celsius como cadena.
celsius_str = input("Introduce la temperatura en Celsius: ")

# Se intenta convertir la cadena a float y calcular Fahrenheit.
try:
    celsius = float(celsius_str)
    fahrenheit = (celsius * 9/5) + 32
    print(f"La temperatura en Fahrenheit es: {fahrenheit}")
except ValueError:
    print("Error: El valor ingresado no es un número válido. Asegúrate de introducir un número.")
```
